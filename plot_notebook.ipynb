```python
import os
import numpy as np
import matplotlib.pyplot as plt
import utils_mpl
```

```python
# set global parameters
utils_mpl.set_global()

# create the output folder
os.makedirs("render", exist_ok=True)
```

```python
# define dummy data
x = [0.0, 1.0, 2.0]
y = [0.0, 1.0, 0.0]

# get the axis ticks
xticks = np.linspace(0.0, 2.0, 5)
yticks = np.linspace(0.0, 1.0, 6)
```

```python
# create a figure with a determined size
(fig, ax) = utils_mpl.get_fig(size=(3.5, 3.0), dpi=200)

# add the histograms
ax.plot(x, y, color="r", lw=1.5)

# set the x-axis limit and format
utils_mpl.set_x_axis(ax, bnd=xticks, margin=0.0, log=False)
utils_mpl.set_format(ax.xaxis, ticks=xticks, fmt="${x:+.1f}$")

# set the y-axis limit and format
utils_mpl.set_y_axis(ax, bnd=yticks, margin=0.1, log=False)
utils_mpl.set_format(ax.yaxis, ticks=yticks, fmt="${x:.1f}$")

# set the legend and labels
ax.set_xlabel("x-axis (unit)")
ax.set_ylabel("y-axis (unit)")
ax.set_title("Plot Title")

# set the grid
utils_mpl.set_grid(fig, ax, major=True, minor=False)
```

```python
# save the plot for Inkscape
utils_mpl.save_svg(fig, "render/jupyter.svg")
```
